Duplicate Rows removed

```python
df = df.drop_duplicates()
```

```python
print(df.duplicated().sum())
```

Null Value find and fill

```python
print(df.isnull().sum())
```

```python
# Numerical columns
numeric_cols = [
    "Age",
    "Study_Hours_Per_Week",
    "Attendance_Percentage",
    "Assignment_Score",
    "Test1_Score",
    "Test2_Score",
    "Final_Exam_Score",
    "Previous_Percentage",
]

for col in numeric_cols:
    df[col] = df[col].fillna(df[col].mean())

# Categorical column
df["City"] = df["City"].fillna(df["City"].mode()[0])
df["Parent_Education"]=df["Parent_Education"].fillna(df["Parent_Education"].mode()[0])
```

```python
print(df.isnull().sum())
```

```python
df.head(5)
```

```python
df["Student_Name"] = df["Student_Name"].str.strip()
```

```python
df["Student_Name"]
```

```python
df["Student_Name"].is_unique
```

```python
df["Age"] = df["Age"].astype(int)
```

```python
df["Age"].dtype
```

```python
df["Age"]
```

```python
text_cols = ["Gender", "City", "Subject"]

for col in text_cols:
    df[col] = df[col].str.strip().str.title()
```

```python
for col in text_cols:
    print(col)
    print(df[col].unique())
```

```python
df.head()
```

```python
df["Parent_Education"] = (
    df["Parent_Education"]
    .str.strip()
    .str.title()
)
```
