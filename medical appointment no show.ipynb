```python
df = df[df['Age']>=0]
```

```python
df['Age'] = df['Age'].abs()
```

```python
df['Age']
```

```python
# PatientId ko int64 mein convert karna
df['PatientId'] = df['PatientId'].astype('int64')
```

```python
df['PatientId']
```

```python
df['ScheduledDay'] = pd.to_datetime(df['ScheduledDay'])
df['AppointmentDay'] = pd.to_datetime(df['AppointmentDay'])
```

```python
df.dtypes
```

```python
df.head()
```

```python
df['Diabetes']=df['Diabetes'].map({0:'No',1:'Yes'})
```

```python
df['Diabetes']
```

```python
# 1. Check karein ki kitni duplicate rows hain
print("Duplicate rows count:", df.duplicated().sum())

# 2. Agar duplicate rows hain, toh unhe drop (remove) kar dein
df.drop_duplicates(inplace=True)

# 3. Verify karne ke liye ki row count kitna bacha
print("Total rows after dropping duplicates:", df.shape[0])
```

```python
df.to_csv('cleaned_medical_appointment_data.csv', index=False)
```
